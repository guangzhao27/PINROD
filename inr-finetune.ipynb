```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision


import optuna

# You can use Matplotlib instead of Plotly for visualization by simply replacing `optuna.visualization.matplotlib` with
# `optuna.visualization.matplotlib.matplotlib` in the following examples.
from optuna.visualization import plot_contour
from optuna.visualization import plot_edf
from optuna.visualization import plot_intermediate_values
from optuna.visualization import plot_optimization_history
from optuna.visualization import plot_parallel_coordinate
from optuna.visualization import plot_param_importances
from optuna.visualization.matplotlib import plot_rank
from optuna.visualization.matplotlib import plot_slice
from optuna.visualization.matplotlib import plot_timeline
```

```python
run_name = 'SO-fine-bs2_sf4_tf2_san1_epoch601'
storage_url = 'sqlite:///'+f'/pscratch/sd/g/gzhao27/INR/SOMA/results/{run_name}-optuna.db'
    
study = optuna.create_study(
                            study_name=run_name,
                            direction='minimize', 
                            sampler=optuna.samplers.TPESampler(), 
                            pruner=optuna.pruners.HyperbandPruner(),
                            storage=storage_url,
                            load_if_exists=True,
                            )
```

```python
import plotly.graph_objects as go
from optuna.trial import TrialState
start, end = 1, 68
filtered_trials = [
    t for t in study.trials 
    if start <= t.number <= end and t.state == TrialState.COMPLETE
]

# Create the custom plot
fig = go.Figure()

for trial in filtered_trials:
    if trial.intermediate_values:  # Ensure trial has intermediate values
        x = list(trial.intermediate_values.keys())  # Steps
        y = list(trial.intermediate_values.values())  # Intermediate values
        fig.add_trace(go.Scatter(x=x, y=y, mode="lines+markers", name=f"Trial {trial.number}"))

# Customize plot layout
fig.update_layout(
    title="Intermediate Values for Trials 50-60",
    xaxis_title="Step",
    yaxis_title="Intermediate Value",
    legend_title="Trials",
)

# Show plot
fig.show()
```

```python
trial_num = 33

import subprocess

# Define the script and arguments
slurm_script = "/pscratch/sd/g/gzhao27/INR/SOMA/script/nersc-soma-inr-finetuning-realrun.sh"
args = [str(x) for x in study.trials[trial_num].params.values()]  # Example arguments

# Submit the script with arguments
result = subprocess.run(
    ["sbatch", slurm_script] + args,  # Append args to the sbatch command
    capture_output=True,
    text=True
)

# Display the result
if result.returncode == 0:
    print("Job submitted successfully:")
    print(result.stdout)
else:
    print("Error submitting job:")
    print(result.stderr)
```
